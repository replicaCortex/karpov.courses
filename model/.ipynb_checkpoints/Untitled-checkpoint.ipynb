```python
import pandas as pd
from sqlalchemy import create_engine


def batch_load_sql(query: str) -> pd.DataFrame:
    CHUNKSIZE = 20000
    engine = create_engine(
        "postgresql://robot-startml-ro:pheiph0hahj1Vaif@postgres.lab.karpov.courses:6432/startml"
    )
    conn = engine.connect().execution_options(stream_results=True)
    chunks = []
    for chunk_dataframe in pd.read_sql(query, conn, chunksize=CHUNKSIZE):
        chunks.append(chunk_dataframe)
    conn.close()
    return pd.concat(chunks, ignore_index=True)
```

```python
df_user_data = batch_load_sql("SELECT * FROM user_data WHERE user_id >= 20000 and user_id <= 30000 LIMIT 10000")
df_post_data = batch_load_sql("SELECT * FROM post LIMIT 10000")
df_feed_data = batch_load_sql("SELECT * FROM feed_data WHERE user_id >= 20000 and user_id <= 30000 LIMIT 50000")
```

```python
df_user_data_1 = pd.get_dummies(df_user_data, columns = ['country', 'city', 'os'])
df_user_data_1 = df_user_data_1.drop(["source"], axis = 1)
```

```python
df_user_data_1 = df_user_data_1.set_index("user_id")
df_user_data_1
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
vect = vectorizer.fit_transform(df_post_data['text'])
```

```python
import numpy as np

tfidf_sums = vect.sum(axis=1)
df_post_data_1 = df_post_data.copy()
df_post_data_1['tfidf_sum'] = np.array(tfidf_sums).flatten()
df_post_data_1 = pd.get_dummies(df_post_data_1, columns = ['topic'])
df_post_data_1 = df_post_data_1.drop(['text'],axis = 1)
```

```python
df_post_data_1 = df_post_data_1.set_index("id")
df_post_data_1
```

```python
df = pd.concat([df_post_data_1.reset_index(drop=True),
                       df_user_data_1.reset_index(drop=True)], axis=1)
```

```python
prob = 1 / (1 + np.exp(-0.5 * df['tfidf_sum'] + 0.3 * df['age'])) 
df['target'] = np.random.binomial(1, prob)
```

```python
df = df.dropna()
df
```

```python
df['target'].value_counts(normalize=True)
```

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline

X = df.drop("target", axis=1)
y = df["target"]

rf_pipe = Pipeline([
    ("scal", MinMaxScaler()),
    ("clf", RandomForestClassifier())
])

gb_pipe = Pipeline([
    ("scal", MinMaxScaler()),
    ("clf", GradientBoostingClassifier())
])

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y)

rf_pipe.fit(X_train, y_train)
gb_pipe.fit(X_train, y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

rf_probs = rf_pipe.predict_proba(X_test)[:, 1]
gb_probs = gb_pipe.predict_proba(X_test)[:, 1]

# ROC curve
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probs)
gb_fpr, gb_tpr, _ = roc_curve(y_test, gb_probs)

plt.figure(figsize=(8,6))
plt.plot(rf_fpr, rf_tpr, label=f'Random Forest (AUC = {roc_auc_score(y_test, rf_probs):.2f})')
plt.plot(gb_fpr, gb_tpr, label=f'Gradient Boosting (AUC = {roc_auc_score(y_test, gb_probs):.2f})')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
```

```python
def recommend_for_user(user_id: int, top_k: int = 5):
    # 1) фичи пользователя
    u = df_user_data_1.loc[[user_id]]
    # 2) блок всех постов
    p = df_post_data_1.reset_index(drop=True)
    p = p.reset_index()
    # 3) повторяем u для всех постов
    users_block = pd.concat([u]*len(p), ignore_index=True)
    # 4) склеиваем
    X_cand = pd.concat([users_block.reset_index(drop=True), p.reset_index(drop=True)], axis=1)
    # 5) приводим к колонкам оригинального X
    X_cand = X_cand[X.columns]
    # 6) предсказываем
    rf_p = rf_pipe.predict_proba(X_cand)[:,1]
    gb_p = gb_pipe.predict_proba(X_cand)[:,1]
    avg_p = (rf_p + gb_p)/2
    # 7) топ-K индексы
    top_idx = np.argsort(avg_p)[::-1][:top_k]
    # 8) возвращаем post_id
    return p.loc[top_idx, 'id'].tolist()

# === 8. Пример вызова ===
user_id = 29566
print("Топ-5 для пользователя", user_id, ":", recommend_for_user(user_id))
```

```python
# Предположим, что в marge_df есть колонки:
# ['post_id','text','topic', … ]

# Считаем TF-IDF-сумму:
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

post_texts = marge_df[['post_id','text','topic', 'user_id']].drop_duplicates('post_id').set_index('post_id')
vect        = TfidfVectorizer()
tfidf_m     = vect.fit_transform(post_texts['text'].fillna(''))
post_texts['tfidf_sum'] = np.array(tfidf_m.sum(axis=1)).flatten()

# One-hot кодируем topic:
post_static = pd.get_dummies(
    post_texts.reset_index(), 
    columns=['topic'], 
).set_index('post_id')


post_static = post_static.drop('text', axis=1)
post_static = post_static.map(lambda x: 1 if x is True else 0 if x is False else x)
```

```python
post_static
```

```python
df = pd.merge(user_static, post_static, left_on="user_id", right_on = "user_id", how="left")
```

```python
df.corr()
```

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline



X = df.drop("target", axis=1)
y = df["target"]

rf_pipe = Pipeline([
    ("scal", MinMaxScaler()),
    ("clf", RandomForestClassifier())
])

gb_pipe = Pipeline([
    ("scal", MinMaxScaler()),
    ("clf", GradientBoostingClassifier())
])

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y)

rf_pipe.fit(X_train, y_train)

import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

rf_probs = rf_pipe.predict_proba(X_test)[:, 1]

# ROC curve
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probs)

plt.figure(figsize=(8,6))
plt.plot(rf_fpr, rf_tpr, label=f'Random Forest (AUC = {roc_auc_score(y_test, rf_probs):.2f})')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
```

```python
import pandas as pd
import numpy as np

def build_X_cand(user_id: int):
    # 1) фичи пользователя
    u = user_static.iloc[[user_id]]                 # DataFrame 1×M
    # 2) блок всех постов
    p = post_static.reset_index(drop=True)         # DataFrame N×K
    p = p.drop(['user_id'], axis=1)
    # 3) дублируем u под каждый пост
    users_block = pd.concat([u]*len(p), ignore_index=True)
    # 4) склеиваем по колонкам
    X_cand = pd.concat([users_block, p], axis=1)
    X_cand = X_cand.drop(['target'], axis=1)
    return X_cand

# пример

X_cand = build_X_cand(3)
print("Размер X_cand:", X_cand.shape)
```
